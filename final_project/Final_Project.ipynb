```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# ——— 参数 & 路径 ——— #
base_dir = "/content/drive/MyDrive/CS445 Project/Final/asset"
img_name = "girl.jpg"         # 可换 "uiuc.jpg" 或 "girl.jpeg"
tex_name = "paper.jpg"        # 水彩纸纹理扫描图
img_path = f"{base_dir}/{img_name}"
tex_path = f"{base_dir}/{tex_name}"

# ——— 1. 读图 & 纹理 ——— #
img = cv2.imread(img_path)
if img is None: raise FileNotFoundError(img_path)
h, w = img.shape[:2]
# 纸张纹理
if Path(tex_path).exists():
    tex = cv2.imread(tex_path, cv2.IMREAD_GRAYSCALE)
    tex = cv2.resize(tex, (w, h))
    tex = cv2.cvtColor(tex, cv2.COLOR_GRAY2BGR) / 255.0
else:
    tex = np.ones_like(img, dtype=np.float32)

# ——— 2. 色块初步柔化（模拟颜料散开） ——— #
# 2.1 中度双边滤波
bleed = cv2.bilateralFilter(img, d=15, sigmaColor=100, sigmaSpace=100)
# 2.2 OpenCV Stylization （增加水彩氛围）
styl = cv2.stylization(bleed, sigma_s=60, sigma_r=0.5)

# ——— 3. 边缘笔触（模拟细笔） ——— #
# 3.1 用 XDoG 取替 Canny：线条更有手绘感
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float32) / 255.0
g1 = cv2.GaussianBlur(gray, (5,5), 0)
g2 = cv2.GaussianBlur(gray, (5,5), 1.0)
dog = g1 - 0.99 * g2
xdog = 1.0 / (1.0 + np.exp(- (dog - 0.1) * 50))
edges = (xdog * 255).astype(np.uint8)
edges = cv2.bitwise_not(edges)  # 白底黑线 → 黑底白线
edges = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR).astype(np.float32) / 255.0

# ——— 4. 流动晕染（Color Flow） ——— #
def flow_blur(src, k=21, dirs=16):
    out = np.zeros_like(src, dtype=np.float32)
    center = k//2
    for i in range(dirs):
        theta = i * np.pi/dirs
        kernel = np.zeros((k,k), np.float32)
        x = int(center + np.cos(theta)*center)
        y = int(center + np.sin(theta)*center)
        cv2.line(kernel, (center,center), (x,y), 1, 1)
        s = kernel.sum() or 1
        kernel /= s
        out += cv2.filter2D(src, -1, kernel)
    return out / dirs

flow = flow_blur(styl.astype(np.float32), k=31, dirs=16)

# ——— 5. 最终融合 ——— #
# styl: 水彩色块 & 柔化， flow: 进一步流动感， edges: 笔触， tex: 纸张
out = styl.astype(np.float32) * 0.7 \
    + flow * 0.2 \
    + edges * 0.05 * 255 \
    + tex * 0.05 * 255
out = np.clip(out, 0, 255).astype(np.uint8)

# ——— 6. 显示对比 ——— #
plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.title("Original")
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2,2,2)
plt.title("Stylization + Bilateral")
plt.imshow(cv2.cvtColor(styl, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2,2,3)
plt.title("XDoG Edges")
plt.imshow(edges[:,:,0], cmap='gray')
plt.axis("off")

plt.subplot(2,2,4)
plt.title("Final Watercolor")
plt.imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# 路径 & 载入
base = "/content/drive/MyDrive/CS445 Project/Final/asset"
img = cv2.imread(f"{base}/girl.jpg")
h, w = img.shape[:2]

# 纸张纹理
tex_path = f"{base}/paper.jpg"
if Path(tex_path).exists():
    tex = cv2.imread(tex_path, cv2.IMREAD_GRAYSCALE)
    tex = cv2.resize(tex, (w, h)) / 255.0
    tex = cv2.cvtColor(tex, cv2.COLOR_GRAY2BGR)
else:
    tex = np.zeros((h,w,3), np.float32)

# 1. 先用 Stylization 强化柔和度，再轻度 Bilateral
styl = cv2.stylization(img, sigma_s=80, sigma_r=0.7)  # ↑ 提高 watercolor 感
bleed = cv2.bilateralFilter(styl, d=9, sigmaColor=50, sigmaSpace=50)

# 2. XDoG 边缘：弱化线条 & 提高阈值，让轮廓更“散”
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float32)/255.
g1 = cv2.GaussianBlur(gray, (7,7), 0)
g2 = cv2.GaussianBlur(gray, (7,7), 1.5)
dog = g1 - 0.95*g2
# 调整阈值和对比度
tau, phi = 0.2, 20
xd = 1/(1+np.exp(-phi*(dog - tau)))
edges = (xd*255).astype(np.uint8)
edges = cv2.bitwise_not(edges)
edges = cv2.GaussianBlur(edges, (5,5), 0)  # 再模糊一次，弱化
edges = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR) / 255.0

# 3. Color Flow：对柔和色块（bleed）做更强的方向性晕染
def flow_blur(src, k=31, dirs=12):
    out = np.zeros_like(src, np.float32)
    c = k//2
    for i in range(dirs):
        th = i*np.pi/dirs
        kern = np.zeros((k,k),np.float32)
        x,y = int(c+np.cos(th)*c), int(c+np.sin(th)*c)
        cv2.line(kern, (c,c),(x,y),1,1)
        kern /= kern.sum() or 1
        out += cv2.filter2D(src, -1, kern)
    return out/dirs

flow = flow_blur(bleed.astype(np.float32), k=31, dirs=12)

# 4. 最终融合：加大 flow 占比，弱化 edges
out = bleed.astype(np.float32)*0.5 \
    + flow*0.35 \
    + edges*0.05*255 \
    + tex*0.1*255
out = np.clip(out,0,255).astype(np.uint8)

# 显示
plt.figure(figsize=(12,8))
plt.subplot(1,2,1)
plt.title("Before (Styl + Bilateral)")
plt.imshow(cv2.cvtColor(bleed, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title("After Tuning → Fluid diffusion ↑ Lines ↓")
plt.imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.tight_layout()
plt.show()
```

## b. Natural Style

這個方法的核心是：

1. 多層 **median filter + bilateral filter** 柔化色塊
2. **Laplacian** 邊緣強化輪廓色階
3. 混合兩者 → 柔順過渡、模擬顏料邊界的擴散


### 🆚 和 Method 1 差異：

| 特徵 | Method 1 (筆觸風格) | Method 2 (柔水彩風格) |
|------|--------------------|----------------------|
| 筆觸輪廓 | 有 XDoG 線條筆觸感 | 幾乎無明顯邊緣線條 |
| 色彩擴散 | 有流動感處理（flow） | 自然平滑的色塊 |
| 紋理 | 可加入紙張紋理 | 本身較乾淨平滑 |
| 整體感覺 | 像手繪插畫 | 像輕淡寫意水彩畫 |
| 效果張力 | 對比強、結構清楚 | 柔和低對比，像暈開 |


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

def method2_watercolor(image_path, save_path="out/method2_result.jpg"):
    # Step 0: 讀圖
    img = cv2.imread(image_path)
    if img is None:
        raise FileNotFoundError("找不到圖片")
    image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Step 1: 多層柔化 → 顏料色塊擴散
    temp = cv2.medianBlur(img, 3)
    smooth = cv2.bilateralFilter(temp, 9, 75, 75)
    for _ in range(2):
        smooth = cv2.bilateralFilter(smooth, 9, 75, 75)

    # Step 2: Laplacian 邊緣模擬水彩邊界
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    lap = cv2.Laplacian(gray, cv2.CV_8U, ksize=3)
    lap_inv = cv2.bitwise_not(lap)
    lap_color = cv2.cvtColor(lap_inv, cv2.COLOR_GRAY2BGR)

    # Step 3: 混合邊緣與柔化圖像
    blended = cv2.addWeighted(smooth.astype(np.float32), 0.95,
                              lap_color.astype(np.float32), 0.05, 0)
    blended = np.clip(blended, 0, 255).astype(np.uint8)

    # Step 4: 顯示
    plt.figure(figsize=(14, 6))
    plt.subplot(1, 3, 1)
    plt.title("Original")
    plt.imshow(image_rgb)
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.title("Smoothed")
    plt.imshow(cv2.cvtColor(smooth, cv2.COLOR_BGR2RGB))
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.title("Method 2 Result (Soft Watercolor)")
    plt.imshow(cv2.cvtColor(blended, cv2.COLOR_BGR2RGB))
    plt.axis('off')

    plt.tight_layout()
    plt.show()

    # Step 5: 儲存
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    cv2.imwrite(save_path, blended)
```

## c. Mixed Style

> **純 OpenCV 實作的 `method1`（進階筆觸水彩） + `method2`（柔和自然水彩）**  
並且可以用 `alpha` 控制兩種風格的融合比例。



## 🎨 Alpha 調整參考：

| alpha | 效果偏向 |
|-------|-----------|
| `1.0` | 完全是 Method 1（有筆觸、有流動） |
| `0.0` | 完全是 Method 2（柔和自然） |
| `0.5` | 各佔一半，結合結構與柔感 |


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from pathlib import Path

def hybrid_watercolor_opencv(image_path, alpha=0.5, save_path=None):

    img = cv2.imread(image_path)
    if img is None:
        raise FileNotFoundError(f"读取失败：{image_path}")
    h, w = img.shape[:2]

    # Method1：Stylization + Bilateral + XDoG + Flow Blur
    styl = cv2.stylization(img, sigma_s=80, sigma_r=0.7)
    bleed = cv2.bilateralFilter(styl, d=9, sigmaColor=50, sigmaSpace=50)


    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float32) / 255.0
    g1 = cv2.GaussianBlur(gray, (7,7), 0)
    g2 = cv2.GaussianBlur(gray, (7,7), 1.5)
    dog = g1 - 0.95 * g2
    tau, phi = 0.2, 20
    xdog = 1.0 / (1.0 + np.exp(-phi * (dog - tau)))
    edges1 = (xdog * 255).astype(np.uint8)
    edges1 = cv2.bitwise_not(edges1)
    edges1 = cv2.GaussianBlur(edges1, (5,5), 0)
    edges1 = cv2.cvtColor(edges1, cv2.COLOR_GRAY2BGR)

    def flow_blur(src, k=31, dirs=12):
        out = np.zeros_like(src, np.float32)
        c = k // 2
        for i in range(dirs):
            theta = i * np.pi / dirs
            kern = np.zeros((k, k), np.float32)
            x = int(c + np.cos(theta) * c)
            y = int(c + np.sin(theta) * c)
            cv2.line(kern, (c, c), (x, y), 1, 1)
            kern /= kern.sum() or 1
            out += cv2.filter2D(src, -1, kern)
        return (out / dirs).astype(np.uint8)

    flow = flow_blur(bleed, k=31, dirs=12)


    method1 = (bleed.astype(np.float32) * 0.5 +
               flow.astype(np.float32)  * 0.35 +
               edges1.astype(np.float32) * 0.1 +
               np.zeros_like(bleed, dtype=np.float32) * 0.05)

    method1_u8 = np.clip(method1, 0, 255).astype(np.uint8)

    # Method2：Median + Multi-Bilateral + Laplacian
    multi = cv2.medianBlur(img, 7)
    for _ in range(3):
        multi = cv2.bilateralFilter(multi, d=9, sigmaColor=75, sigmaSpace=75)

    gray2 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    lap = cv2.Laplacian(gray2, cv2.CV_8U, ksize=3)
    lap_inv = cv2.bitwise_not(lap)
    edges2 = cv2.cvtColor(lap_inv, cv2.COLOR_GRAY2BGR)

    method2 = cv2.addWeighted(multi.astype(np.float32), 0.9,
                              edges2.astype(np.float32),  0.1, 0)
    method2_u8 = np.clip(method2, 0, 255).astype(np.uint8)

    # Hybrid（same dtype） #
    hybrid = cv2.addWeighted(method1_u8, alpha, method2_u8, 1 - alpha, 0)
    hybrid = np.clip(hybrid, 0, 255).astype(np.uint8)

    plt.figure(figsize=(12,6))
    plt.subplot(1,3,1)
    plt.title("Method1 (Brushy)")
    plt.imshow(cv2.cvtColor(method1_u8, cv2.COLOR_BGR2RGB)); plt.axis('off')
    plt.subplot(1,3,2)
    plt.title("Method2 (Soft)")
    plt.imshow(cv2.cvtColor(method2_u8, cv2.COLOR_BGR2RGB)); plt.axis('off')
    plt.subplot(1,3,3)
    plt.title(f"Hybrid α={alpha}")
    plt.imshow(cv2.cvtColor(hybrid, cv2.COLOR_BGR2RGB)); plt.axis('off')
    plt.tight_layout()
    plt.show()


    if save_path:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        cv2.imwrite(save_path, hybrid)

    return hybrid

```

# 2. Neural Style Transfer (NST)

**Processing Pipeline:**

* Extracts multi-level style/content features from VGG

* Computes style loss (Gram matrix) and content loss

* Optimizes the image iteratively via backpropagation

**Benefits:**

* Can replicate arbitrary painting styles

* Produces expressive, high-quality renderings

* Strong alignment with style image, but limited tunability



```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import os
from tqdm import tqdm
from PIL import Image

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import torchvision.models as models
from torchvision.utils import save_image
```

```python
#hyperparams

num_steps = 1000
learning_rate = 1e-2
alpha = 1
beta = 0.1
IMG_SIZE = 300 if torch.cuda.is_available() else 100

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("cuda" if torch.cuda.is_available() else "cpu")
```

```python
# utils

def load_img_as_tensor(img_name):
    img = cv2.imread(img_name)
    assert isinstance(img,np.ndarray) #To check if image exist

    img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

    img = torch.from_numpy(img/255)
    img = img.unsqueeze(0) # img.shape (IMG_SIZE,IMG_SIZE,3)->(1,IMG_SIZE,IMG_SIZE,3)

    img = torch.einsum("bhwc->bchw",img)
    return img.float()


directory="out"
if not os.path.exists(directory):
    os.makedirs(directory)      # creting out folder
```

```python
# model

class VGG(nn.Module):
    def __init__(self):
        super().__init__()
        self.features_layers = [0, 5, 10, 19, 28]

        self.model = models.vgg19(pretrained=True).eval()

        self.model = self.model.features[:29]

    def forward(self, x):
        features = []

        for idx, layer in enumerate(self.model):
            x = layer(x)

            if idx in self.features_layers:
                features.append(x)

        return features
```

```python
#Loss Function

def gram_matrix(x):
    batch, channels, hight, width = x.shape
    features = x.view(batch * channels, hight * width)
    G = torch.mm(features, features.t())
    return G


class StyleLoss(nn.Module):
    def __init__(self, features):
        super().__init__()
        self.features = [out.detach() for out in features ]

        #precomputing gram matrix for features
        self.target_A = [gram_matrix(fe) for fe in self.features ]


    def forward(self, x):
        loss = 0
        for A, pred in zip(self.target_A, x):
            G = gram_matrix(pred)
            loss += F.mse_loss(A, G)
        return loss


class ContentLoss(nn.Module):
    def __init__(self, features):
        super().__init__()
        self.features = [out.detach() for out in features]

    def forward(self, x):
        loss = 0
        for target, pred in zip(self.features, x):
            loss += F.mse_loss(target, pred)
        return loss
```

```python
def transfer(model, img, style, content, num_steps=500, output_path='out/NST_result.jpg'):
    img = img.detach().clone().to(device).requires_grad_()
    optimizer = optim.Adam([img])

    style_feature = model(style)
    content_feature = model(content)

    content_loss = ContentLoss(content_feature)
    style_loss = StyleLoss(style_feature)

    for idx in tqdm(range(num_steps)):
        img_feature = model(img)

        c_loss = content_loss(img_feature)
        s_loss = style_loss(img_feature)

        loss = alpha * s_loss + beta * c_loss

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    save_image(img.clone().detach(), f'{output_path}')
    return
```

```python
def run_cross_nst(target_paths, style_paths, model, output_prefix="NST_landscape"):
    os.makedirs("out", exist_ok=True)
    results = []

    for ti, target_path in enumerate(target_paths):
        target_tensor = load_img_as_tensor(target_path).to(device)

        for si, style_path in enumerate(style_paths):
            style_tensor = load_img_as_tensor(style_path).to(device)
            img = target_tensor.clone()

            result_name = f"{output_prefix}_T{ti+1}_S{si+1}.jpg"
            output_path = f"out/{result_name}"

            transfer(model, img, style_tensor, target_tensor, num_steps=500, output_path=output_path)

            result_img = Image.open(output_path)
            results.append((f"T{ti+1} + S{si+1}", result_img))

    # Show result
    fig, axs = plt.subplots(2, 2, figsize=(12, 10))
    for i, (title, img) in enumerate(results):
        ax = axs[i // 2][i % 2]
        ax.imshow(img)
        ax.axis('off')
        ax.set_title(f'NST result: {title}')
    plt.tight_layout()
    plt.show()
```

# 3. Hybrid Style: OpenCV Mixed + NST

To overcome the limitations of OpenCV-only and NST-only approaches, we design a hybrid pipeline:

* Use the **Mixed OpenCV** output as the content image

* Apply **NST** with a chosen style painting

* Retain both painterly structure and deep artistic textures

## Why Hybrid?
* OpenCV-only filters offer **fine-grained style control**, but may lack realism

* NST provides **deep artistic stylization**, but is often unstable or too sensitive to the style image

* Combining both gives us **customizability and creativity**, a form of controllable neural style transfer

This hybrid method lets users start with a preferred watercolor base (via alpha-tuned OpenCV filters) and enrich it using the neural style derived from a reference painting—offering the best of both worlds.

## Adjustable alpha parameter:

* alpha = 1.0 → illustrative (sharp lines)

* alpha = 0.0 → natural (soft edges)

* alpha = 0.5 → mixed



```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import os
from tqdm import tqdm
from PIL import Image

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import torchvision.models as models
from torchvision.utils import save_image
```

```python
#hyperparams

num_steps = 1000
learning_rate = 1e-2
alpha = 1
beta = 0.1
IMG_SIZE = 300 if torch.cuda.is_available() else 100

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("cuda" if torch.cuda.is_available() else "cpu")
```

```python
def flow_blur(src, k=31, dirs=12):
    out = np.zeros_like(src, np.float32)
    c = k // 2
    for i in range(dirs):
        theta = i * np.pi / dirs
        kern = np.zeros((k, k), np.float32)
        x = int(c + np.cos(theta) * c)
        y = int(c + np.sin(theta) * c)
        cv2.line(kern, (c, c), (x, y), 1, 1)
        kern /= kern.sum() or 1
        out += cv2.filter2D(src, -1, kern)
    return out / dirs

def preprocess_content_for_hybrid(content_path, alpha=0.5):
    assert 0.0 <= alpha <= 1.0, "alpha must be between 0 and 1"

    img = cv2.imread(content_path)
    if img is None:
        raise FileNotFoundError("cannot find image")
    h, w = img.shape[:2]

    # Illustrative Style
    # Step 1: Stylization + Bilateral
    styl = cv2.stylization(img, sigma_s=80, sigma_r=0.7)
    bleed = cv2.bilateralFilter(styl, d=9, sigmaColor=50, sigmaSpace=50)

    # Step 2: XDoG edges
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float32) / 255.0
    g1 = cv2.GaussianBlur(gray, (7, 7), 0)
    g2 = cv2.GaussianBlur(gray, (7, 7), 1.5)
    dog = g1 - 0.95 * g2
    tau, phi = 0.2, 20
    xdog = 1 / (1 + np.exp(-phi * (dog - tau)))
    edges = (xdog * 255).astype(np.uint8)
    edges = cv2.bitwise_not(edges)
    edges = cv2.GaussianBlur(edges, (5, 5), 0)
    edges = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR).astype(np.float32) / 255.0

    # Step 3: Color Flow
    flow = flow_blur(bleed.astype(np.float32), k=31, dirs=12)

    # Step 4: Texture (optional: replace with blank if no paper)
    tex = np.ones_like(flow, dtype=np.float32) * 0.9

    method1 = (
        bleed.astype(np.float32) * 0.5 +
        flow * 0.35 +
        edges * 0.05 * 255 +
        tex * 0.1 * 255
    )

    # Natural Style: Laplacian + Multi-filter
    img_filtered = cv2.medianBlur(img, 5)
    for _ in range(2):
        img_filtered = cv2.bilateralFilter(img_filtered, d=9, sigmaColor=75, sigmaSpace=75)

    gray2 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    lap = cv2.Laplacian(gray2, cv2.CV_8U, ksize=3)
    lap_color = cv2.cvtColor(255 - lap, cv2.COLOR_GRAY2BGR)

    method2 = cv2.addWeighted(img_filtered.astype(np.float32), 0.95, lap_color.astype(np.float32), 0.05, 0)

    # Blend both
    hybrid = cv2.addWeighted(method1, alpha, method2, 1 - alpha, 0)
    hybrid = np.clip(hybrid, 0, 255).astype(np.uint8)

    os.makedirs("out", exist_ok=True)
    hybrid_path = "out/hybrid_content.jpg"
    cv2.imwrite(hybrid_path, hybrid)
    return hybrid_path
```

```python
def load_img_as_tensor(img_name):
    img = cv2.imread(img_name)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) / 255.0
    img = torch.from_numpy(img).permute(2, 0, 1).unsqueeze(0)
    return img.float().to(device)
```

```python
def run_hybrid_style_transfer(content_path, style_path, model, output_name="out/Hybrid_result.jpg"):
    content_tensor = load_img_as_tensor(content_path)
    style_tensor = load_img_as_tensor(style_path)
    img = content_tensor.clone()

    transfer(model, img, style_tensor, content_tensor)

    result = Image.open(output_name)
    return result
```

```python
# model

class VGG(nn.Module):
    def __init__(self):
        super().__init__()
        self.features_layers = [0, 5, 10, 19, 28]

        self.model = models.vgg19(pretrained=True).eval()

        self.model = self.model.features[:29]

    def forward(self, x):
        features = []

        for idx, layer in enumerate(self.model):
            x = layer(x)

            if idx in self.features_layers:
                features.append(x)

        return features
```

```python
#Loss Function

def gram_matrix(x):
    batch, channels, hight, width = x.shape
    features = x.view(batch * channels, hight * width)
    G = torch.mm(features, features.t())
    return G


class StyleLoss(nn.Module):
    def __init__(self, features):
        super().__init__()
        self.features = [out.detach() for out in features ]

        #precomputing gram matrix for features
        self.target_A = [gram_matrix(fe) for fe in self.features ]


    def forward(self, x):
        loss = 0
        for A, pred in zip(self.target_A, x):
            G = gram_matrix(pred)
            loss += F.mse_loss(A, G)
        return loss


class ContentLoss(nn.Module):
    def __init__(self, features):
        super().__init__()
        self.features = [out.detach() for out in features]

    def forward(self, x):
        loss = 0
        for target, pred in zip(self.features, x):
            loss += F.mse_loss(target, pred)
        return loss
```

```python
def transfer(model, img, style, content, num_steps=500, output_name="out/Hybrid_result.jpg"):
    img = img.detach().clone().to(device).requires_grad_()
    optimizer = optim.Adam([img])

    style_feature = model(style)
    content_feature = model(content)

    content_loss = ContentLoss(content_feature)
    style_loss = StyleLoss(style_feature)

    for idx in tqdm(range(num_steps)):
        img_feature = model(img)

        c_loss = content_loss(img_feature)
        s_loss = style_loss(img_feature)

        loss = alpha * s_loss + beta * c_loss

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    save_image(img.clone().detach(), output_name)
```

```python
def run_all_hybrid_cross_styles(target_paths, style_paths, model):
    alphas = [0.0, 0.5, 1.0]
    hybrid_imgs = []
    hybrid_labels = []
    result_imgs = []
    result_labels = []

    for target_path in target_paths:
        base_name = os.path.splitext(os.path.basename(target_path))[0]
        for alpha in alphas:
            hybrid_path = preprocess_content_for_hybrid(target_path, alpha=alpha)
            img = cv2.imread(hybrid_path)
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            hybrid_imgs.append(img)
            hybrid_labels.append(f'{base_name} α={alpha}')

            # NST
            for style_path in style_paths:
                style_name = os.path.splitext(os.path.basename(style_path))[0]
                content_tensor = load_img_as_tensor(hybrid_path)
                style_tensor = load_img_as_tensor(style_path)
                img_tensor = content_tensor.clone()

                output_name = f"out/NST_{base_name}_alpha{alpha}_{style_name}.jpg"
                transfer(model, img_tensor, style_tensor, content_tensor, output_name=output_name)
                result_img = Image.open(output_name)
                result_imgs.append(result_img)
                result_labels.append(f'{base_name} α={alpha} → {style_name}')

    # Show hybrid preprocessed images（6 images）
    fig, axs = plt.subplots(2, 3, figsize=(15, 8))
    for i, (img, label) in enumerate(zip(hybrid_imgs, hybrid_labels)):
        axs[i // 3][i % 3].imshow(img)
        axs[i // 3][i % 3].axis('off')
        axs[i // 3][i % 3].set_title(label)
    plt.suptitle("Hybrid Preprocessed Images", fontsize=16)
    plt.tight_layout()
    plt.show()

    # Show NST result（12 images）
    fig, axs = plt.subplots(3, 4, figsize=(20, 12))
    for i, (img, label) in enumerate(zip(result_imgs, result_labels)):
        axs[i // 4][i % 4].imshow(img)
        axs[i // 4][i % 4].axis('off')
        axs[i // 4][i % 4].set_title(label, fontsize=10)
    plt.suptitle("NST Results from Hybrid Preprocessed Images", fontsize=18)
    plt.tight_layout()
    plt.show()
```
